```python
# How many participants?
print("Participants:", data["Subj_idx"].nunique())

# Which tasks, and how many trials in each?
print(data["Task"].value_counts())

# Which confidence values appear? (We expect 0, 10, 20 ... 100)
print(sorted(data["Confidence"].dropna().unique()))

# How many missing values in each column?
print(data.isna().sum())
```

```python
# How are Stimulus and Response coded? (We expect the same two values, e.g. 1 and 2)
print(data["Stimulus"].unique(), data["Response"].unique())

# Make a Correct column: 1 if the response matches the stimulus, 0 if not
data["Correct"] = (data["Stimulus"] == data["Response"]).astype(int)

# Structure check only: how many confidence ratings are below 50% (values 0-4)?
print("Ratings below 50%:", (data["Confidence"] < 5).sum(), "of", len(data))
```
